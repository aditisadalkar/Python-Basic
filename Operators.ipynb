```python
#Arithmetic Operator with integers
a=20
b=2
print(a+b)
print(a-b)
print(a*b)
print(a/b) #(always float)
print(a//b) #(floor division)
print(a%b) #modual division(baki division /remainder
print(a**b)
```

```python
#Arithmetic Operator with floats
a=20.5
b=2
print('a+b =',a + b)
print('a-b =',a - b)
print('a*b =',a * b)
print('a/b =',a / b)
print('a//b =',a // b)
print('a%b =',a % b)
print('a**b =',a ** b)
```

```python
#Division behavior
print(20/2)
print(20//2)
print(20.0/2)
print(20.0//2)
```

```python
#String operator with and *
print("anvi"+"20")
print(2 * "anvi")
print("anvi"* 2)
```

```python
#Relational (comparison) Operators with numbers
a=20
b=30
print(a>b)
print(a>=b)
print(a<b)
print(a<=b)
```

```python
#Relational Operators with string
a="anvi"
b="anvi"
print(a>b)
print(a>=b)
print(a<b)
print(a<=b)
```

```python
print(len(a))
print(len(b))
```

```python
#Relational Operators with booleans
print(True > True)
print(True >= True)
print(20 > True)
print(False > True)
```

```python
# Chaining of Relational Operators 
print(20<30)
print(20<30<40)
print(20<30<40<50)
print(20<30<40<50>60)
```

```python
#Equality operator:== and !=
print(20 == 30)
print(20 != 30)
print(20 == True)    # False  (True ==1)
print(False == False)
print("anvi" == "anvi")
print(20 == "anvi")   #False  (no error - compatible types not required)

#Chaining equality
print(20 == 30 == 40 == 50)
print(20 == 20 == 20 == 20)
```

```python
#Logical Operator with boolean
print(True and False)
print(True or False)
print( not False)
```

```python
#Logical operator with non-boolean types
print(10 and 20)
print(0 and 20)

#x or y
print(10 or 20)
print(0 or 20)

#not x
print(not 10)
print(not 0)
```

```python
#Logical operator with string
print("anvi" and "anvisoft")
print("" and "anvi")
print("anvi" and "")
print("" or "anvi")
print(not "")
print(not "anvi")
```

```python
#Assignment operator
a=20
print('Initial a:',a)

a+=5
print('a+=5 ->',a)
a-=3
print('a-=3 ->',a)
a*=2
print('a*=2 ->',a)
a/=4
print('a/=4 ->',a)
a//=2
print('a//=2 ->',a)
a=20
a%=3
print('a%=3 ->',a)
a=3
a**=3
print('a**=3 ->',a)
```

```python
#Identity operators
a=30
b=30
print(a is b)
```

```python
x = True
y = True
print(x is y)
```

```python
#Identity operator-string
a="umangi"
b="umangi"
print(id(a))
print(id(b))
print(a is b)
```

```python
#Identity operator -list
list1=["one","two","three"]
list2=["one","two","three"]
print(id(list1))
print(id(list2))
print(list1 is list2)
print(list1 is not list2)
print(list1 == list2)
```
