# Immutable and Mutable 

```python
#Immutable and Mutable 
x = 'anvi'
print(id(x))
print(type(x),x)
```

```python
x = 'umangi'
print(id(x))
print(x)
```

```python
lst=['aditi','shravani']
print(id(lst))
print(type(lst),lst)
```

```python
lst.append('umangi')
print(lst,type(lst),id(lst))
```

```python
lst=[20,50,70]
print("Before:",id(lst))
lst.append(40)
print("After:",id(lst))
print(lst)
```

# Dict

```python
#Dict
student={"name":"Aditi"}
print("Before:",id(student))
student["age"]=19
print("After:",id(student))
print(student)
```

# Set(Mutable)

```python
#Set(Mutable)
s={20,30,40}
print("Before:",id(s))
s.add(50)
print("After:",id(s))  
print(s)
```

# Integer(Immutable)

```python
#Integer(Immutable)
x=20
print("Before:",id(x))
x=x+5
print("After:",id(x))  
print(x)
```

```python
#Tuple(Immutable)
t=(10,30,40)
print("Before:",id(t))
t=t+(50,)
print("After:",id(t))
print(t)
```
