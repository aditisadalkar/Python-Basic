```python
# Factorial using loop
num = int(input("enter num"))
def fact(num):
    result = 1
    while num >= 1:
        result = result * num
        num = num - 1
fact(num)
for i in range(1,6):
    print("factorial of {i} is: {fact(i)}")
```

Types of Arguments

```python
# 1.Positional argument
def sub(a, b):
    print(a - b)
sub(100,300)
sub(300,100)
```

```python
def sub(a, b):
    print(a + b)
sub(100,200)
sub(200,100)
```

```python
# 2.Keyword argument
def wish(name,msg):
    print("hello", name, msg)
wish("good morning", "anvi")
wish(msg="good morning", name="durga")
```

```python
wish(msg="durga", name="good morning")
```

```python
# 3.Default argument
def wish(name="guest"):
    print("hello",name,"good morning")
wish("durga")
wish()
```

```python
# 4.Variable length argument(*args)
def sum_all(*n):
    total = 0
    for n1 in n:
        total += n1
    print("sum =",total)
sum_all()
sum_all(20)
sum_all(20,30)
sum_all(20,30,40)
```

```python
def add_all(*numbers):
    total = 0

    for num in numbers:
        total += num
    return total
print(add_all())
print(add_all(10))
print(add_all(10,20))
print(add_all(10,20,30,40))
```

```python
def find_max(*numbers):
    print("tuple:",numbers)

    if numbers:
        print("maximum =",max(numbers))
    else:
        print("no numbers provided")
find_max(10,2,40,25)
find_max(30,44,50,75,87,99)
```

```python
# why only tuple
def add(*numbers):
    c = 0
    print("first number:",numbers[0])

    for i in numbers:
        c += i
    print(f"sum is {c}")
add(20,25,15,4)    
```

Mix positional + *args


```python
def f1(n1, *s):
    print("first arg:",n1)
    print("rest:",s)

f1(10)
f1(10,20,30,40)
f1(10,"A",30,"B")
```

```python
def student_marks(name, *marks):
    print("student:",name)
    print("marks:",marks)
student_marks(75,80,85,20,98,"aditi")
```

```python
# *args before a named arg
def f1(*s, n1):
    for s1 in s:
        print(s1)
    print("n1 = ",n1)

f1("A", "B", n1=10)

```

kwargs

```python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, "=", value)

student_info(name="aditi", branch="CO", year=3)
```

```python
# 4b.keyword variable length argument(**kwargs)    
# Internally stored as a DICTIONARY
def display(**kwargs):
    for k, v in kwargs.items():
        print(k, "=", v)
        
print("---- call 1 ---")
display(n1=20, n2=30, n3=40)

print("--- call 2 ---")
display(rno=111, name="anvi", marks=83, subject="python")

```

```python
# --- case study:mixing arg types ---
def f(arg1,arg2,arg3=5,arg4=8):
    print(arg1,arg2,arg3,arg4)

f(4,2)
f(20,30,40,50)
f(35,50, arg4=100)
f(arg4=3,arg1=3,arg2=5)
```

args and *kwargs in python

```python
def info_person(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:",kwargs)

info_person(1,2, name="anvi", age=16, dept="CSE")
```
