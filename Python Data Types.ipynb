# Accessing character of string

```python
# 1.By using index
s = 'anvi'
print(s[0])
print(s[3])
print(s[-1])

# Accessing out-of-range index
try:
    print(s[10])
except IndexError as e:
    print("IndexError:",e) 
```

```python
# Program: Display character by positive and negative index
s = input("enter some string:")
i = 0
for x in s:
    print("the character present at positive index {} and at negative index {} is {}".format(i, i-len(s),(x)))
    i += 1
```

```python
# Accessing Character using slice operator
s = "learning python is very very easy!!!"
print(s[1:7:1])
print(s[1:7])
print(s[1:7:2])
print(s[:7])
print(s[7:])
print(s[::])
print(s[::-1])
```

```python
# Mathamatical operator for string
# 1.len() built-in function
s = 'anvi'
print(len(s))
print(s)
print(id(s),type(s))
```

# Important function of list

```python

# len()-number of element
n = [10,20,30,40]
print(len(n))
```

```python
# count()
n = [1,3,3,3,3,4,4]
print(n.count(1))
print(n.count(3))
print(n.count(4))
print(n.count(5))
```

```python
# index()
n = [1,3,3,3,3,4,4]
print(n.index(1))
print(n.index(3))
print(n.index(4))
```

```python
# append()- add item at the end
lst = []
lst.append("A")
lst.append("B")
lst.append("C")
print(lst)
```

```python
# append - add multiples of 10 up to 100
lst = []
for i in range(0,101,10):
    if i % 10 == 0:
        lst.append(i)
print(lst)        
    
```

```python
lst = []
for i in range(0,101,10):
    lst.append(i)  
print(lst)        
    
```

```python
# insert() 
n = [1,2,3,4,5]
n.insert(1,999)
print(n)
```

```python
# Beyond range insert at end
n = [1,2,3,4,5]
n.insert(10,888)
n.insert(-10,999)
print(n)
```

```python
# extend() - add all items from another list
order1 = ["chicken", "mutton", "fish"]
order2 = ["RC", "KF", "FO"]
order1.extend(order2)
print(order1)
```

```python
order = ["chicken", "mutton", "fish"]
order.extend(["mushroom","panner"])
print(order)
```

```python
# remove()
n = [20,30,40,50]
n.remove(20)
print(n)
```

```python
# pop()
n = [20,30,40,50]
print(n.pop())
print(n.pop())
print(n)
```

```python
# pop(index)
n = [20,30,40,50,60,70]
print(n.pop())
print(n.pop(1))
print(n)
```

```python
# Ordering function
#1.reverse()
n = [10,20,30,40,50]
n.reverse()
print(n)
```

```python
#sort()
n = [20,30,5,6,1,4,0]
n.sort()
print(n)
```

```python
s = ["Dog", "Banana", "Cat", "Apple"]
s.sort()
print(s)
```
