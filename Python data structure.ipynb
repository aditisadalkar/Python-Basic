1) Tuple Data Structure

```python
# Parantheses are optional
t = 20,30,40,50
print(t)
print(type(t))
```

```python
# Empty tuple
t = (8,)
print(type(t))
```

```python
#without trailing comma - NOT Tuple
t = (10,)
print(t)
print(type(t))
```

```python
#without trailing comma - IS Tuple
t = (10,)
print(t)
print(type(t))
```

Creating Tuple Object

```python
# 1.Empty tuple
t = ()
print()

# 2.Single-valued
t = (10,)
print(t)

# 3.Multi-valued
t = (10,30,40)
print(t)

# Without parentheses
t = 10,30,40
print(t)
```

```python
# 4.Using tuple() function
lst = [10,30,40]
t = tuple(lst)
print(t)

t = tuple(range(20,30,2))
print(t)
```

Accessing element of a tuple

```python
# 1. By index
t = (10,20,30,40,50)
print(t[0])
print(t[1])
print(t[2])
print(t[3])
print(t[-2])
```

```python
# 2. Slice operator
t = (10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[::3])
```

Mathamatical operator for tuple

```python
# concatnation (+)
t1 = (20,30,40)
t2 = (50,60,70)
t3 = t1 + t2
print(t3)

# Repetition (*)
t1 = (20,30,40)
t2 = t1 * 3
print(t2)
```

Important function of tuple

```python
# len()-number of element
t = (20,30,40,50)
print(len(t))

# count()-occurrences of an element
t = (20,30,20,20,0)
print(t.count(20))

# index()-first occurrences index
print(t.index(20))
```

```python
# sorted()-returns a sorted LIST(orignal tuple unchanged)
t = (40,20,30,10)
t1 = sorted(t)
print(t1)

# reverse sort 
t1 = sorted(t, reverse=True)
print(t1,type(t1))
```

```python
# min() and max()
t = (40,10,30,20)
print(min(t))
print(max(t))
```

Tuple packing and unpaking


```python
# Tuple packing
a = 20
b = 30
c = 40
d = 50
t = a,b,c,d
print(t)
```

```python
# Tuple unpaking
t = (10,20,30,40)
a,b,c,d = t
print("a=",a, "b=",b ,"c=",c , "d=",d)
print(type(a))
```

Tuple Comprehesion

```python
t = (x**2 for x in range(1,6))
print(type(t))
print(t)

for x in t:
    print(x)
```

Sum and average of a tuple

```python
t = eval(input("enter tuple of numbers:"))
l = len(t)
total = 0
for x in t:
    total = total + x
print("the sum=",total)
print("the average=",total / l)
print(l)
```

2) Set Data Structure

```python
#creating set object
s = {10,20,30,40}
print(s,type(s))
```

```python
# from a list(removes duplicate automatically)
l = [10,20,30,40,10,20,10]
s = set(l)
print(s)

# from range
s = set(range(5))
print(s)
```

```python
# {} create a dict ,not a set
s = {}
print(s,type(s))

# use set() for empty set
s = set()
print(s,type(s))
```

```python
# important function of set
#1.add()
s = {20,30,40}
s.add(50)
print(s)
```

```python
#2.update()-add multiple value
s = {10,20,30}
l = {40,50,60,10}
s.update(l,range(5))
print(s)
```

```python
#add() vs update()
s = {10,20}
#s.add(10,20,30)
#s.update(10)

s.update(range(1,10,2),range(0,10,2))
print(s)
```

```python
s = {10,20}
s.update([10,20,30,88,99,100,400])
s
```

```python
#copy()
s = {10,20,30}
s1 = s.copy()
print(s1)
print(s)
```

```python
# pop()
s = {40,10,20,30}
print(s)
print(s.pop())
print(s)
```

```python
#remove()
s = {40,10,30,20}
s.remove(10)
print(s)
```

```python
# discard()
s = {10,20,30}
s.discard(10)
print(s)
s.discard(50)
print(s)
```

```python
# clear()
s = {10,20,30}
print(s)
s.clear()
print(s)
```
