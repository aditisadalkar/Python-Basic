```python
#if statement
name = input("enter name:")
if name == "umangi":
    print("hello umangi good morning")
print("how are you !!") 
```

```python
# if-else statement
name = input("enter name")
if name == "neha":
    print("hello neha good morning")
else:
    print("hello guest good morning")
print("how are you !!")    
```

```python
# if-else statement
password = input("enter name:")
if password == "neha":
    print("you loggedn in")
else:
    print("denied")

```

```python
# if-elif-else statement
marks = 75
if marks >= 90:
    print("grade A")
elif marks >= 60:
    print("grade B")
else:
    print("grade C")
```

```python
# while loop  print number from 1 to 10
x = 1
while x <=10:
    print(x)
    x =x+1

```

```python
#sum of first n numbers
n=int(input("enter number:"))
total =0
i=1
while i <=n:
    total = total + i
    i=i+1
print("the sum of first", n, "number is:", total)    
```

```python
3)Transfer statement
```

```python
#break statement  when i==7
for i in range(11):
    if i==7:
        print("processing is enough.. please break")
        break
    print(i)   
```

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)    


```

```python
#continue
for i in range(1, 6):
    if i == 3:
        continue
    print(i)    
```

```python
#pass
for i in range(1, 4):
    pass
print("loop completed")    
```
